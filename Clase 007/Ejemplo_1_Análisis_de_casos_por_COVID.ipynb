```python
print(df[df.index==77724])

print(df[df.index==77725])

print(df[df.index==77726])
```

```python

# 🩹 Completar valores faltantes con la media entre el valor anterior y el siguiente
df['Confirmed'] = df['Confirmed'].fillna(df['Confirmed'].interpolate())
print(df[df.index==77725]) #otra forma de  obtener lo mismo es df.iloc[[77725]]
# ✅ vemos que el cambio se aplicó correctamente
```

```python
print(df[df.index==19778])

print(df[df.index==19779])

print(df[df.index==19780])
```

```python

# 🩹 Completar valores faltantes con la media entre el valor anterior y el siguiente
df['Recovered'] = df['Recovered'].fillna(df['Recovered'].interpolate())
print(df[df.index==19779])
```

```python
# 🩹 Se hace lo mismo con Deaths

df['Deaths'] = df['Deaths'].fillna(df['Deaths'].interpolate())
```

Tengan en cuenta que este enfoque asume que los valores en la columnas  están ordenados cronológicamente y por país. Si los valores no están ordenados adecuadamente, se debe ordenar el DataFrame  antes de aplicar la interpolación.



ejemplo df.sort_values(by=["Country", "Date"])

> 💡 **¿Por qué interpolar y no usar `fillna(0)`?**
> Como se trata de una serie temporal acumulada (casos confirmados/recuperados/muertes por día), poner `0` en un hueco generaría un salto irreal en el gráfico. La interpolación estima el valor faltante a partir del dato anterior y el siguiente, respetando la tendencia real de la curva.

```python
fig=px.choropleth(df, locations="Country", locationmode="country names", color= "Confirmed",animation_frame="Date")

# 🗺️ con Plotly Express (px) creamos un gráfico de mapa animado (choropleth map).
# primero se especifica el df
# locations: Especifica la columna del DataFrame que se utilizará para asignar los datos a ubicaciones geográficas en el mapa.
# locationmode: Especifica el modo de ubicación utilizado para interpretar los valores de ubicación. En este caso, se utiliza "country names" para indicar que los valores de ubicación en la columna "Country" corresponden a los nombres de los países.
# color: Especifica la columna del DataFrame que se utilizará para determinar el color de cada ubicación en el mapa. Aquí, se utiliza la columna "Confirmed" para representar la cantidad de casos confirmados de alguna variable
# animation_frame: Especifica la columna del DataFrame que se utilizará para crear una animación en el gráfico. En este caso, se utiliza la columna "Date"

# se recomienda revisar https://plotly.github.io/plotly.py-docs/generated/plotly.express.choropleth.html
```

```python
fig.update_layout(title="Mapa de casos confirmados de COVID-19 en el mundo")
fig.show()

# 👀 fijense como se va moviendo la escala de confirmados con el tiempo
```

```python


fig = px.choropleth(

    df,

    locations="Country",

    locationmode="country names",

    color="Confirmed",

    range_color=[0,1000000], #es clave esta medida si queremos que nuestra escala no cambie

    animation_frame="Date",

    color_continuous_scale="Viridis"  # Establece la escala de colores aquí

)



fig.update_layout(

    coloraxis_colorbar=dict(

        title="Confirmed Cases"  # Cambia el título de la barra de colores si lo deseas

    )

)





fig.show()








```

```python
fig.update_layout(title="Mapa de casos confirmados de COVID-19 en el mundo")

fig.show()
```

## 🌎 En el ejemplo anterior vimos la evolución de casos por COVID-19 a nivel mundial. Sin embargo, choropleth permite hacer un análisis más detallado por continente o para USA, mediante el parámetro "scope"

```python
fig=px.choropleth(df, locations="Country", scope= "south america",locationmode="country names", color= "Confirmed",animation_frame="Date")

fig.update_layout(title="Mapa de casos confirmados por COVID-19 en América del Sur")

fig.show()
```

### 📍 Ahora hacemos la visualización mediante un scatter-Geoplot

```python
fig=px.scatter_geo(df, locations="Country", locationmode="country names", color="Confirmed",size="Confirmed", hover_name="Country", animation_frame="Date",title="Evolución en el tiempo")

fig.update(layout_coloraxis_showscale=False,layout_template="plotly_dark") #cambiamos el fondo por negro y actualizamos la figura

# Ajustar la velocidad de la animación

fig.layout.updatemenus[0].buttons[0].args[1]["frame"]["duration"] = 500

fig.show()
```

## 🏆 Top 10 países con más casos confirmados

Reforzamos `groupby` + `sort_values` de pandas para armar un ranking simple.

```python
top10 = df.groupby("Country")["Confirmed"].max().sort_values(ascending=False).head(10)

fig_top10 = px.bar(top10, x=top10.values, y=top10.index, orientation="h",
                    labels={"x": "Casos confirmados", "y": "País"},
                    title="🏆 Top 10 países con más casos confirmados de COVID-19",
                    color=top10.values, color_continuous_scale="Oranges")
fig_top10.update_layout(yaxis={'categoryorder':'total ascending'})
fig_top10.show()
```
