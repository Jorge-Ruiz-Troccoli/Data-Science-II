```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'key': ['b', 'b', 'a', 'c', 'a', 'a', 'b'],
 'first': range(7)})
df1
```

```python
df2 = pd.DataFrame({'key': ['a', 'b', 'd'], 'second': range(3)})
df2
```

```python
df3=pd.merge(left=df1,right=df2, on ='key',how='left')
df3
```

Al tener un valor faltante una columna númerica por defecto pasa a ser float, reemplacemos por 999 como valor faltante y luego la convertimos en int.

```python
import numpy as np
df3["second"].replace(np.nan, 999, inplace=True)
df3["second"]=df3["second"].astype(int)
df3
```

```python
#aplicamos un filtro para obtener un .describe()
#~ tilde de la Ñ o virgulilla

filtro=df3.second==999
df3.loc[~filtro]
#df3[~(df3.second==999)] otra forma de hacer lo mismo

```

# Ejemplo en vivo Parte 2

```python
clase0=pd.read_excel('clase0.xlsx')
clase0
```

```python
clase1=pd.read_excel('clase1.xlsx')
clase1
```

```python
clase1.groupby(by=['key1','key2']).mean()
# clase1.groupby(['key1','key2']).mean() se puede usar sin el by
```

```python

"""
La función pd.merge() en pandas se utiliza para combinar dos DataFrames en función de una
o más columnas comunes, especificadas mediante el parámetro on.
En este ejemplo, estamos realizando un merge entre los DataFrames clase0 y clase1 utilizando
las columnas 'key1' y 'key2' como claves de combinación.
"""
pd.merge(clase0, clase1, on =['key1','key2'])
# por defecto el "how"  es inner
```

```python
pd.merge(clase0,clase1, on =['key1','key2'],how= 'inner')

#igual al anterior pero siendo explícitos en el "how"
```

```python
"""
El parámetro how='outer' en la función pd.merge() indica que se realizará una combinación externa o
"outer join" entre los DataFrames que estás uniendo.
Esta opción de combinación incluye todas las filas de ambos DataFrames,
incluso aquellas que no tienen una coincidencia en las columnas de combinación especificadas.
"""

pd.merge(clase0,clase1, on =['key1','key2'],how= 'outer')
```

```python
"""
El parámetro how='left' en la función pd.merge() indica que se realizará una combinación izquierda o "left join"
entre los DataFrames que estás uniendo. En un left join, el resultado contendrá todas las filas del DataFrame de la izquierda
(el primer DataFrame en la función pd.merge()) y las filas coincidentes del DataFrame de la derecha (el segundo DataFrame en la función).

Si alguna fila del DataFrame de la izquierda no tiene una coincidencia en las columnas de combinación,
las columnas correspondientes del DataFrame de la derecha se llenarán con NaN (valores nulos).
"""



pd.merge(clase0,clase1, on =['key1','key2'],how= 'left')
```

```python
"""
El parámetro how='right' en la función pd.merge() indica que se realizará una combinación derecha o "right join"
entre los DataFrames que estás uniendo. En un right join, el resultado contendrá todas las filas del DataFrame de la derecha
 y las filas coincidentes del DataFrame de la izquierda.

Si alguna fila del DataFrame de la derecha no tiene una coincidencia en las columnas de combinación,
las columnas correspondientes del DataFrame de la izquierda se llenarán con NaN (valores nulos).
"""


pd.merge(clase0,clase1, on =['key1','key2'],how= 'right')
```

```python
# ahora veamos .concat()

"""pd.concat([c1, c2, c3, c4], axis=0) combina los DataFrames c1, c2, c3 y c4 uno debajo del otro,
creando un nuevo DataFrame que contiene todas las filas de los DataFrames originales de forma consecutiva.

El resultado tendrá la misma cantidad de columnas que los DataFrames originales y contendrá todas las filas de c1,
seguidas de todas las filas de c2, luego todas las filas de c3 y finalmente todas las filas de c4.

Es importante que los DataFrames deben tener las mismas columnas en el mismo orden para realizar una concatenación vertical adecuada."""


c1=pd.merge(clase0,clase1, on =['key1','key2'],how= 'right')
c2=pd.merge(clase0,clase1, on =['key1','key2'],how= 'left')
c3=pd.merge(clase0,clase1, on =['key1','key2'],how= 'outer')
c4=pd.merge(clase0,clase1, on =['key1','key2'],how= 'inner')
pd.concat([c1,c2,c3,c4],axis=0)
```

```python
c1
```

```python
#probemos que pasa si eliminamos la columna data del df c1
c1.drop("data", axis=1, inplace=True)
c1
```

```python
pd.concat([c1,c2,c3,c4],axis=0)
#podemos ver que en data c1 va aparecer ahora valores NaN
```

```python
"""
el parámetro axis=1 en pandas se utiliza para indicar que se aplicará el método
a lo largo del eje horizontal (columnas) en lugar del eje vertical (filas).
"""

a=pd.concat([c1,c2],axis=1)
#a=pd.concat([c1,c2],axis="columns")
a
```
