```python

lista_paises_sur=['Argentina','Chile','Colombia','Bolivia','Peru','Brazil','Uruguay','Venezuela','Paraguay','Ecuador']
lista_paises_nor=['United States','Canada','Mexico']


def promedio_paises(lista_paises):
    lista_valores=[]
    for pais in lista_paises:
        pais=dft[pais].astype("str")
        pais= pais.apply(transformar)
        pais.index=pais.index.astype(int)
        promedio_pais=pais.loc[2015:2021].mean()
        lista_valores.append(promedio_pais)

    suma = sum(lista_valores)
    promedio = suma/len(lista_valores)
    return f"El valor promedio entre los {len(lista_paises)} paises es de {round(promedio,2)}"



```

```python
# el peor camino usando for al final para el promedio
lista_paises_sur=['Argentina','Chile','Colombia','Bolivia','Peru','Brazil','Uruguay','Venezuela','Paraguay','Ecuador']
lista_paises_nor=['United States','Canada','Mexico']


def promedio_paises(lista_paises):
    lista_valores=[]
    suma=0
    for pais in lista_paises:
        pais=dft[pais].astype("str")
        pais= pais.apply(transformar)
        pais.index=pais.index.astype(int)
        promedio_pais=pais.loc[2015:2021].mean()
        lista_valores.append(promedio_pais)

    for elementos in lista_valores:
        suma+=elementos
    promedio=suma/len(lista_valores)


    return f"El valor promedio entre los {len(lista_paises)} paises es de {round(promedio,2)}"
```

```python
df_copia=df1

variables= df1.columns

for i in range(len(variables)):
    if df_copia[variables[i]].isna().sum()>len(df_copia)*0.1: #mayor al 10% de valores faltantes se elimina la columna
        df_copia.drop(variables[i], axis=1, inplace=True)
df_copia
```

```python
percent_missing = df_copia.isnull().sum() * 100 / len(df_copia)
missing_value_df_copia = pd.DataFrame({'Columnas': df_copia.columns,
                                       "total": df_copia.isnull().sum(),
                                 'Porcentaje_Missing': percent_missing})
missing_value_df_copia
```
